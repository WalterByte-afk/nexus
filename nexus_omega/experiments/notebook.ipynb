# NEXUS-Ω Experiment Notebook

This notebook demonstrates the NEXUS-Ω brain-inspired architecture.

```python
import torch
import sys
sys.path.insert(0, '..')

from nexus_omega.core.omega_system import create_nexus_omega
```

```python
# Create the model
model = create_nexus_omega("small", vocab_size=10000)
print(f"Model created with {model.count_parameters()['total']:,} parameters")
```

```python
# Run inference
input_ids = torch.randint(0, 10000, (2, 32))
output = model(input_ids)
print(f"Logits shape: {output.logits.shape}")
```

```python
# Enable online learning
output = model(input_ids, enable_online_learning=True)
print(f"Online learning enabled")
print(f"Plastic weight updates: {model.blocks[0].dual_plasticity.update_count}")
```

```python
# Run consolidation
report = model.consolidate()
print(f"Consolidation report: {report}")
```

```python
# Memory footprint
print(f"Memory footprint: {model.get_memory_footprint()}")
```
